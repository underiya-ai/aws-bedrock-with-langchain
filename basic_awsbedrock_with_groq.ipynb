```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq

load_dotenv()

llm =ChatGroq(
               model="openai/gpt-oss-120b",
               api_key=os.getenv("GROQ_API_KEY")
             )

query = "Can you assist to me for one week only ?"
res = llm.invoke(query)

print(res.content)

```
